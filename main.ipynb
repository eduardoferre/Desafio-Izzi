```python
import pandas as pd

df = pd.read_csv('BaseDeDados1.csv')

df
```

```python

print(df[["Faturamento Mensal (R$)", "Custo Operacional (R$)"]].head())
```

```python
df["Faturamento Mensal (R$)"] = df["Faturamento Mensal (R$)"].astype(str).str.strip()
df["Custo Operacional (R$)"] = df["Custo Operacional (R$)"].astype(str).str.strip()
df["Impostos Pagos (R$)"] = df["Impostos Pagos (R$)"].astype(str).str.strip()

df["Faturamento Mensal (R$)"] = df["Faturamento Mensal (R$)"].replace({'R\$': '', '\.': '', ',': '.'}, regex=True)
df["Custo Operacional (R$)"] = df["Custo Operacional (R$)"].replace({'R\$': '', '\.': '', ',': '.'}, regex=True)
df["Impostos Pagos (R$)"] = df["Impostos Pagos (R$)"].replace({'R\$': '', '\.': '', ',': '.'}, regex=True)

df["Faturamento Mensal (R$)"] = pd.to_numeric(df["Faturamento Mensal (R$)"], errors='coerce')
df["Custo Operacional (R$)"] = pd.to_numeric(df["Custo Operacional (R$)"], errors='coerce')
df["Impostos Pagos (R$)"] = pd.to_numeric(df["Impostos Pagos (R$)"], errors='coerce')

df
```

```python
df["Lucro Operacional (R$)"] = df["Faturamento Mensal (R$)"] - df["Custo Operacional (R$)"]

df["Margem Operacional (%)"] = ((df["Lucro Operacional (R$)"] / df["Faturamento Mensal (R$)"]) * 100).round(2)

df["Percentual de Impostos (%)"] = ((df["Impostos Pagos (R$)"] / df["Faturamento Mensal (R$)"]) * 100).round(2)

df["Rentabilidade Líquida (R$)"] = df["Lucro Operacional (R$)"] - df["Impostos Pagos (R$)"]
df["Rentabilidade Liquida (%)"] = ((df["Lucro Operacional (R$)"] - df["Impostos Pagos (R$)"]) / df["Faturamento Mensal (R$)"] * 100).round(2)



df
```

```python
clientes_maior_margem = df.sort_values(by="Margem Operacional (%)", ascending=False).head(5)

clientes_maior_margem
```

```python
clientes_mais_impostos = df.sort_values(by="Percentual de Impostos (%)", ascending=False).head(5)

clientes_mais_impostos
```

```python
clientes_rentabilidade = df.sort_values(by="Rentabilidade Líquida (R$)", ascending=False)

clientes_rentabilidade
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))
plt.bar(df['Cliente'], df['Margem Operacional (%)'], color='skyblue')
plt.title('Margem Operacional dos Clientes')
plt.xlabel('Clientes')
plt.ylabel('Margem Operacional (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('margem_operacional.png')
plt.close()

plt.figure(figsize=(10, 6))
plt.bar(df['Cliente'], df['Percentual de Impostos (%)'], color='salmon')
plt.title('Percentual de Impostos sobre o Faturamento')
plt.xlabel('Clientes')
plt.ylabel('% Impostos sobre Faturamento')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('impostos_percentual.png')
plt.close()

plt.figure(figsize=(10, 6))
plt.bar(df['Cliente'], df['Rentabilidade Líquida (R$)'], color='lightgreen')
plt.title('Rentabilidade Líquida dos Clientes')
plt.xlabel('Clientes')
plt.ylabel('Rentabilidade Líquida (R$)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('rentabilidade_liquida.png')
plt.close()

plt.figure(figsize=(10, 6))
plt.bar(df['Cliente'], df['Rentabilidade Liquida (%)'], color='orange')
plt.title('Rentabilidade Líquida (%) dos Clientes')
plt.xlabel('Clientes')
plt.ylabel('Rentabilidade Líquida (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('rentabilidade_liquida_percentual.png')
plt.close()
```

```python
relatorio = """
Relatório de Análise Financeira dos Clientes

1. Clientes com maior margem operacional:
{}
 
2. Clientes que pagam mais impostos proporcionalmente ao faturamento:
{}

3. Ranking dos clientes mais rentáveis (Lucro Operacional - Impostos):
{}

""".format(clientes_maior_margem[["Cliente", "Margem Operacional (%)"]].to_string(index=False),
           clientes_mais_impostos[["Cliente", "Percentual de Impostos (%)"]].to_string(index=False),
           clientes_rentabilidade[["Cliente", "Rentabilidade Líquida (R$)"]].to_string(index=False))

print(relatorio)
```

```python
from fpdf import FPDF

pdf = FPDF()
pdf.add_page()

pdf.set_font("Arial", size=16, style='B')
pdf.cell(200, 10, txt="Relatório de Análise Financeira dos Clientes", ln=True, align='C')

pdf.ln(10) 

pdf.set_font("Arial", size=12)
pdf.multi_cell(0, 10, relatorio)

pdf.ln(10) 

pdf.image('margem_operacional.png', x=10, w=180)
pdf.ln(90)  

pdf.image('impostos_percentual.png', x=10, w=180)
pdf.ln(90) 

pdf.image('rentabilidade_liquida.png', x=10, w=180)
pdf.ln(90)

pdf.image('rentabilidade_liquida_percentual.png', x=10, w=180)

pdf.output("relatorio_financeiro.pdf")

```

```python
import plotly.graph_objects as go
import plotly.express as px

df_sorted_margem = df.sort_values(by='Margem Operacional (%)', ascending=False)
df_sorted_impostos = df.sort_values(by='Percentual de Impostos (%)', ascending=False)
df_sorted_rentabilidade = df.sort_values(by='Rentabilidade Líquida (R$)', ascending=False)
df_sorted_rentabilidade_percentual = df.sort_values(by='Rentabilidade Liquida (%)', ascending=False)

fig1 = go.Figure(data=[go.Bar(
    x=df_sorted_margem['Cliente'],
    y=df_sorted_margem['Margem Operacional (%)'],
    text=df_sorted_margem['Margem Operacional (%)'],
    textposition='auto',  
    marker_color='skyblue'
)])

fig1.update_layout(
    title="Margem Operacional dos Clientes (%)",
    xaxis_title="Clientes",
    yaxis_title="Margem Operacional (%)",
    xaxis_tickangle=-45,
    showlegend=False,
    template="plotly_dark"  
)

fig1.show()

fig2 = go.Figure(data=[go.Bar(
    x=df_sorted_impostos['Cliente'],
    y=df_sorted_impostos['Percentual de Impostos (%)'],
    text=df_sorted_impostos['Percentual de Impostos (%)'],
    textposition='auto',  
    marker_color='salmon'
)])

fig2.update_layout(
    title="Percentual de Impostos sobre o Faturamento (%)",
    xaxis_title="Clientes",
    yaxis_title="% Impostos sobre Faturamento",
    xaxis_tickangle=-45,
    showlegend=False,
    template="plotly_dark"  
)

fig2.show()

fig3 = go.Figure(data=[go.Bar(
    x=df_sorted_rentabilidade['Cliente'],
    y=df_sorted_rentabilidade['Rentabilidade Líquida (R$)'],
    text=df_sorted_rentabilidade['Rentabilidade Líquida (R$)'],
    textposition='auto', 
    marker_color='lightgreen',
    texttemplate='%{text:.2f}'  
)])

fig3.update_layout(
    title="Rentabilidade Líquida dos Clientes (R$)",
    xaxis_title="Clientes",
    yaxis_title="Rentabilidade Líquida (R$)",
    xaxis_tickangle=-45,
    showlegend=False,
    template="plotly_dark" 
)

fig3.show()

fig4 = go.Figure(data=[go.Bar(
    x=df_sorted_rentabilidade_percentual['Cliente'],
    y=df_sorted_rentabilidade_percentual['Rentabilidade Liquida (%)'],
    text=df_sorted_rentabilidade_percentual['Rentabilidade Liquida (%)'],
    textposition='auto', 
)])

fig4.update_layout(
    title="Rentabilidade Liquida (%) dos Clientes",
    xaxis_title="Clientes",
    yaxis_title="Rentabilidade Liquida (%)",
    xaxis_tickangle=-45,
    showlegend=False,
    template="plotly_dark"  
)

fig4.show()
```
